```python

import numpy as np

arr=np.array([1,2,3,4,5,6,7,2,3,6,8,5,3,2,0])

print("mean:" , np.mean(arr))

print("median:" ,np.median(arr))

values, counts = np.unique(arr, return_counts=True)
mode = values[np.argmax(counts)]

print("Mode:", mode)

print("standard deviation:" , np.std(arr))

print("variance:" , np.var(arr))

```

```python
#without function

#mean calculation
def mean_calculation(list_):
  sum=0
  count=0
  for number in list_:
    sum+=number
    count+=1
  return (sum/count)

mean_ans=mean_calculation([1,2,3,4,5,6,7])

print(mean_ans)

#median
def median_calculation(list_):
  count=0
  for number in list_:
    count+=1
  list_.sort()
  if((count//2)*2==count):
    return (list_[count//2]+list_[(count//2)-1])/2
  else:
    return list_[count//2]

median_ans_odd=median_calculation([1,2,3,4,5,6,7])
median_ans_even=median_calculation([1,2,3,4,5,6])

print(median_ans_odd)
print(median_ans_even)

#mode
def mode_calculation(list_):
    frequency = {}

    for number in list_:
        if number in frequency:
            frequency[number] += 1
        else:
            frequency[number] = 1

    # Find the maximum frequency
    mode = max(frequency , key = frequency.get)
    return mode

print(mode_calculation([1,2,3,1,2,2,7]))

#standard deviation
def standard_deviation(list_):
  mean_=mean_calculation(list_)
  temp=0
  for number in list_:
    temp+=(number-mean_)**2
  return (temp/(len(list_)-1))**0.5

sd_ans=standard_deviation([1,2,3,4,5,6,7])

print(sd_ans)

# variance
def variance(list_):
  var = standard_deviation(list_)
  return var**2

var_ans=variance([1,2,3,4,5,6,7])

print(var_ans)
```
