```python
import numpy as np

data = np.array([10, 20, 30, 40, 50], dtype=float)

min_val = np.min(data)
max_val = np.max(data)

min_max = (data - min_val) / (max_val - min_val)

mean = np.mean(data)
std = np.std(data)

z_score = (data - mean) / std


j = len(str(int(np.max(np.abs(data)))))
decimal_scaling = data / (10 ** j)


print("Original Data: ", data)
print("Min-Max: ", min_max)
print("Z-Score: ", z_score)
print("Decimal Scaling: ", decimal_scaling)
```

# chi-square

```python
import numpy as np
from sklearn.feature_selection import chi2


X = np.array([
    [10, 20],
    [20, 10],
    [15, 25],
    [25, 15],
    [30, 20],
    [20, 30]
])
y = np.array([0, 1, 0, 1, 1, 0])

chi_values, p_values = chi2(X, y)

print("Chi-Square values:", chi_values)
print("P-values:", p_values)
```
