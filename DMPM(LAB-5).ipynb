```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# 1. Transaction data
transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Butter'],
    ['Milk', 'Bread', 'Butter']
]

# 2. Get all unique items
items = sorted(set(
    item
    for transaction in transactions
    for item in transaction
))

# 3. One-hot encode transactions
data = []

for transaction in transactions:
    row = {
        item: (item in transaction)
        for item in items
    }
    data.append(row)

df = pd.DataFrame(data)

print("Transaction Data:")
print(df)

# 4. Find frequent itemsets
frequent_itemsets = apriori(
    df,
    min_support=0.4,
    use_colnames=True
)

print("\nFrequent Itemsets:")
print(frequent_itemsets)

# 5. Generate association rules
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.7
)

# 6. Display important columns
print("\nAssociation Rules:")
print(
    rules[
        ['antecedents',
         'consequents',
         'support',
         'confidence',
         'lift']
    ]
)
```
