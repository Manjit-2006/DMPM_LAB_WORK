```python
# K-Means and K-Medoids Clustering
# Small manually created dataset

import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn_extra.cluster import KMedoids

# --------------------------------------------------
# 1. Create Dataset
# --------------------------------------------------

# Income is in ₹ thousands

data = {
    "Age": [22, 25, 28, 45, 48, 50, 65, 68, 70, 72],
    "Income": [20, 25, 30, 55, 60, 65, 80, 85, 90, 95]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)


# --------------------------------------------------
# 2. Select Features
# --------------------------------------------------

X = df[["Age", "Income"]]


# --------------------------------------------------
# 3. Scale the Data
# --------------------------------------------------

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


# ==================================================
# K-MEANS
# ==================================================

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Train the model
kmeans.fit(X_scaled)

# Add cluster labels
df["KMeans_Cluster"] = kmeans.labels_

# Convert centroids back to original scale
centroids = scaler.inverse_transform(
    kmeans.cluster_centers_
)

print("\n========== K-MEANS ==========")

print("\nClustered Dataset:")
print(df[["Age", "Income", "KMeans_Cluster"]])

print("\nK-Means Centroids:")

for i, center in enumerate(centroids):
    print(
        f"Cluster {i}: "
        f"Age = {center[0]:.2f}, "
        f"Income = ₹{center[1]:.2f} thousand"
    )


# ==================================================
# K-MEDOIDS
# ==================================================

kmedoids = KMedoids(
    n_clusters=3,
    random_state=42,
    method="alternate"
)

# Train the model
kmedoids.fit(X_scaled)

# Add cluster labels
df["KMedoids_Cluster"] = kmedoids.labels_

# Get medoid indices
medoid_indices = kmedoids.medoid_indices_

# Get actual medoid records
medoids = df.iloc[medoid_indices]

print("\n========== K-MEDOIDS ==========")

print("\nClustered Dataset:")
print(df[["Age", "Income", "KMedoids_Cluster"]])

print("\nK-Medoids:")
print(medoids[["Age", "Income"]])
```
